```python
# Exercise 1: Prime Numbers

def check_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    for i in range(2,n):
        if n % i == 0:
            return False
    return True
            

check_prime(89)
```

```python
# Exercise 2: Product of Random Numbers

import random

def product_Rnum():

    num1 = random.randint(1,50)
    num2 = random.randint(1,50)

    correct_ans = num1 * num2

    print(f"What is {num1} x {num2}?")

    user_ans = int(input("Your answer: "))

    if user_ans == correct_ans:
        print(f"Your answer is correct, {user_ans}")
    else:
        print(f"Your answer is incorrect, correct answer is {correct_ans}")

product_Rnum()
```

```python
# Exercise 3: Squares of Even/Odd Numbers

def square_of_even_numbers():
    
    StartNum = 100
    EndNum = 201

    for num in range(StartNum,EndNum):
        if num % 2 == 0:
            print(f"{num*num}")

square_of_even_numbers()
```

```python
# Exercise 4:

def hospital_billing(stay,test_count,med_charg):

    RoomCharges = 2000
    Consultation = 1500
    TestCharges = 300
    Discount = 0

    Total_Bill = (stay*RoomCharges) + (test_count*TestCharges) + med_charg + Consultation

    if Total_Bill > 10000:
        Discount = Total_Bill * 0.10

    print(f"Hospital Bill\n Doctor Consultation Fee: {Consultation}\n Room Charges: {stay*RoomCharges}\n Lab Test Charges: {test_count*TestCharges}\n Medicine Charges: {med_charg}\n Discount: {Discount}\n Total Bill: {Total_Bill-Discount}")

hospital_billing(4,3,2400)
```
