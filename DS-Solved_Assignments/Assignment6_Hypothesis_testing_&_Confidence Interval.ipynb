```python
#Hypothesis testing
```

```python
import math
from scipy.stats import norm

# Given values
sample_mean = 3050
n = 25
mu_x = 600
sigma_x = 25

# Theoretical mean weekly cost
mu = 1000 + (5 * mu_x)

# Standard deviation of weekly cost
sigma = 5 * sigma_x

# Calculate Z statistic
z = (sample_mean - mu) / (sigma / math.sqrt(n))

# Critical value at 5% significance level
alpha = 0.05
critical_value = norm.ppf(1 - alpha)

# P-value
p_value = 1 - norm.cdf(z)

print("Theoretical Mean:", mu)
print("Standard Deviation:", sigma)
print("Z-statistic:", z)
print("Critical Value:", critical_value)
print("P-value:", p_value)

if z > critical_value:
    print("Reject H0: Operating cost has increased.")
else:
    print("Fail to Reject H0: No evidence that operating cost has increased.")
```

```python
#Confidence Intervals
```

```python
import numpy as np
from scipy.stats import t, norm

# Data
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85,
        1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
mean = np.mean(data)
s = np.std(data, ddof=1)

# 99% Confidence Interval using sample standard deviation (t-distribution)
alpha = 0.01
t_value = t.ppf(1 - alpha/2, df=n-1)

margin_t = t_value * (s / np.sqrt(n))
ci_t = (mean - margin_t, mean + margin_t)

print("Sample Mean:", round(mean, 4))
print("Sample Std Dev:", round(s, 4))
print("99% CI using t-distribution:", tuple(round(x, 4) for x in ci_t))


# 99% Confidence Interval using known population standard deviation
sigma = 0.2
z_value = norm.ppf(1 - alpha/2)

margin_z = z_value * (sigma / np.sqrt(n))
ci_z = (mean - margin_z, mean + margin_z)

print("99% CI using known population SD:", tuple(round(x, 4) for x in ci_z))
```

```python
#a. Using sample standard deviation:
#Since the sample size is small (n = 15) and the population standard deviation is unknown, we use the t-distribution.
#
#99% CI = (1.0975, 1.3799) million characters
#
#b. Using known population standard deviation:
#Since the population standard deviation is known (σ = 0.2), we use the standard normal (Z) distribution.
#
#99% CI = (1.1054, 1.3720) million characters.
```
