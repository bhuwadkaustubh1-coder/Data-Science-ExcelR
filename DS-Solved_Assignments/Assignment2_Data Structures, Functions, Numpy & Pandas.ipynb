```python
# 5

def AdultCheck(age : int):
    if age >= 18:
        print("Person is eligible to Vote")
    else:
        print("Person is not eligible to Vote")

AdultCheck(18)
AdultCheck(16)
```

```python
# 6

def result(marks : int):
    if marks >= 90:
        print("A")
    elif marks >= 75:
        print("B")
    elif marks >= 50:
        print("C")
    else:
        print("Fail")

result(40)
result(55)
result(75)
result(91)
        
```

```python
# 7

def check_number(num):
    if num > 0:
        if num % 2 == 0:
            print(f"{num} is positive and even")
        else:
            print(f"{num} is positive but not even")
    elif num == 0:
        print("The number is zero")
    else:
        print(f"{num} is negative")

check_number(100)
```
