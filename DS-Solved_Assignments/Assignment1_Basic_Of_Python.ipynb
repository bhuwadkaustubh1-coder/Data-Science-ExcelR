```python
import pandas as pd

data = pd.read_csv(r"C:\Users\Admin\OneDrive\Desktop\Data Science ExcelR\Assignments\Assignment 3\sales_data_with_discounts.csv")
salesData = data.copy()
salesData.head()
```

```python
salesData.describe()

# describe method will help in identifying numerical columns and its mean, median, mode & std Deviation
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.barplot(data = salesData, x = 'Brand', y = 'Net Sales Value', errorbar = None)
plt.show()
```

```python
salesData['Volume'].plot.hist(bins = 10)
plt.show
# 'Volume' histogram is positively squed
# most of people buy 1 to 5 products
```

```python
salesData['Avg Price'].plot.hist(bins = 10)
plt.show()

# 'Avg Price' histogram is positively squed
# average buying is in most cases 0 to 10000
```

```python
salesData['Discount Amount'].plot.hist(bins = 10)
plt.show()

# 'Discount Amount' histogram is positively squed
# Maximum discount givent for small amount in 0 to 2500
```

```python
salesData['Discount Rate (%)'].plot.hist(bins = 10)
plt.show()

# Discount Rate is negatively squed
# Discount rate is high in most of cases
# most of people are getting discount rate of 17 to 18 %
```

```python
salesData['Net Sales Value'].plot.hist()
plt.show()

# 'Net Sales Value' is positively squied
# Most volume is 0 to 25000
```

```python
salesData['Total Sales Value'].plot.hist(bins =20)
plt.show()

# sales data is positively squed
# Most of volume is in 0 to 25000
```
