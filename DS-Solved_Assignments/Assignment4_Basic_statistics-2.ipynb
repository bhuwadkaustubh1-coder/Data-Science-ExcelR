```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
patient_file = r"C:\Users\Admin\OneDrive\Desktop\Data Science ExcelR\Assignments\Assignment 4\Patient_Data.csv"

patient_df = pd.read_csv(patient_file)
patient_df.head()
```

```python
billing_file = r"C:\Users\Admin\OneDrive\Desktop\Data Science ExcelR\Assignments\Assignment 4\Billing_Data.csv"

billing_df = pd.read_csv(billing_file)
billing_df.head()
```

```python
print("Patient Dataset Information:")
patient_df.info()
```

```python
print("\nBilling Dataset Information:")
billing_df.info()
```

```python
print("Patient Dataset Shape:", patient_df.shape)
print("Billing Dataset Shape:", billing_df.shape)
```

```python
billing_patient_df = patient_df[
    ['PatientID', 'Department', 'Doctor', 'BillAmount']
]

billing_patient_df
```

```python
clean_patient_df = patient_df.copy()

clean_patient_df = clean_patient_df.drop(
    columns=['ReceptionistID', 'CheckInTime']
)

clean_patient_df
```

```python
#Total Bill Amount Per Department

department_billing = (
    clean_patient_df
    .groupby('Department')['BillAmount']
    .sum()
    .reset_index()
)

department_billing
```

```python
patient_counts = patient_df['PatientID'].value_counts()

print(patient_counts)

# Remove duplicate patients based on PatientID

unique_patient_df = clean_patient_df.drop_duplicates(
    subset='PatientID',
    keep='first'
)

unique_patient_df

print("Records before removing duplicates:", len(clean_patient_df))
print("Records after removing duplicates:", len(unique_patient_df))
```

```python
print("Missing BillAmount values:")
print(unique_patient_df['BillAmount'].isnull().sum())

mean_bill = unique_patient_df['BillAmount'].mean()

print("Mean Bill Amount:", mean_bill)

```

```python
unique_patient_df['BillAmount'] = (
    unique_patient_df['BillAmount']
    .fillna(mean_bill)
)

unique_patient_df

print(
    "Missing BillAmount after filling:",
    unique_patient_df['BillAmount'].isnull().sum()
)
```

```python
#Merge Billing Dataset With Patient Dataset

merged_df = pd.merge(
    unique_patient_df,
    billing_df,
    on='PatientID',
    how='left'
)

merged_df

print("Merged Dataset Shape:", merged_df.shape)

print("\nMerged Dataset:")
display(merged_df)
```

```python
print("Missing values in merged dataset:")
print(merged_df.isnull().sum())
```

```python
#Concatenate New Patients Row-Wise

new_patients = pd.DataFrame({
    'PatientID': [107, 108],
    'Name': ['George', 'Helen'],
    'Department': ['Cardiology', 'Neurology'],
    'Doctor': ['Dr. Sharma', 'Dr. Patel'],
    'BillAmount': [4500, 3800]
})

new_patients

updated_patient_df = pd.concat(
    [unique_patient_df, new_patients],
    ignore_index=True
)

updated_patient_df

print("Original Patient Records:", len(unique_patient_df))
print("New Patient Records:", len(new_patients))
print("Updated Patient Records:", len(updated_patient_df))
```

```python
#Concatenate New Billing Columns Column-Wise

billing_columns = billing_df[
    ['InsuranceCovered', 'FinalAmount']
]

billing_columns

patient_billing_combined = pd.concat(
    [
        unique_patient_df.reset_index(drop=True),
        billing_columns.reset_index(drop=True)
    ],
    axis=1
)

patient_billing_combined
```

```python
#Final Cleaned Dataset

print("Final Cleaned Dataset:")
display(merged_df)
```
